```python
#Importing the libraries needed
import numpy as np
import pandas as pd
import difflib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
#data preprocessing
movies = pd.read_csv('movies.csv')
```

```python
movies.shape
```

```python
movies.head()
```

```python
# selecting the relevant features for recommendation

selected_features = ['genres','keywords','tagline','cast','director']
print(selected_features)
```

```python
# replacing the null valuess with null string

for feature in selected_features:
  movies[feature] = movies[feature].fillna('')
```

```python
# combining all the 5 selected features

combined_features = movies['genres']+' '+movies['keywords']+' '+movies['tagline']+' '+movies['cast']+' '+movies['director']
```

```python
print(combined_features)
```

```python
#selected_features = movies[['genres', 'keywords', 'tagline', 'cast', 'director']]
```

```python
#selected_features.head()
```

```python
#selected_features.isnull().sum() #selected_features.fillna("", inplace = True) to replace null values with null string if there are missing values
```

```python
vectorizer = TfidfVectorizer()
```

```python
feature_vectors = vectorizer.fit_transform(combined_features)
```

Cosine Similarity

```python
# getting the similarity scores using cosine similarity
similarity = cosine_similarity(feature_vectors)
```

```python
print(similarity)
```

```python
print(similarity.shape)
```

```python
#getting the movie name
movie_name = input('Enter your favourite movie name:')
```

```python
#creating a list with all the movie names given in the dataset
titles_list = movies['title'].tolist()
```

```python
#finding close match for the given movie name
close_match = difflib.get_close_matches(movie_name, titles_list)
print(close_match)
```

```python
closest_match = close_match[0]
print(closest_match)
```

```python
#finding the index of the movie title
movie_index = movies[movies.title == closest_match]['index'].values[0]
print(movie_index)
```

```python
#getting list of similar movies
similarity_score = list(enumerate(similarity[movie_index]))
#print(similarity_score)
```

```python
len(similarity_score)
```

```python
#sorting based on highestto lowest similarity score
sorted_similar_movies = sorted(similarity_score, key = lambda x:x[1], reverse = True)
#print(sorted_similar_movies)
```

```python
#print the name of similar movies based on the index
print('Movies suggested for you: \n')
i = 1
for movie in sorted_similar_movies:
    index = movie[0]
    title_from_index = movies[movies.index == index]['title'].values[0]
    if i <= 30:
        print(i, ".", title_from_index)
        i += 1
    else:
        break
```
